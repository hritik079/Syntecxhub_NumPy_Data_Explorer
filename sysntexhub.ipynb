```python
import numpy as np
import time

arr = np.array([10, 20, 30, 40, 50])
print("Original Array:", arr)

print("First element:", arr[0])
print("Last element:", arr[-1])
print("Slice (1:4):", arr[1:4])

print("Array + 5:", arr + 5)
print("Array * 2:", arr * 2)

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Max:", np.max(arr))
print("Min:", np.min(arr))
print("Sum:", np.sum(arr))

arr2 = np.arange(1, 10)
matrix = arr2.reshape(3,3)
print("Reshaped Matrix:\n", matrix)

print("Broadcasting (matrix + 10):\n", matrix + 10)

np.save("data_array.npy", arr)
loaded_array = np.load("data_array.npy")
print("Loaded Array:", loaded_array)

list_data = list(range(1000000))
start = time.time()
result_list = [x + 5 for x in list_data]
end = time.time()
print("Python List Time:", end - start)

numpy_data = np.arange(1000000)
start = time.time()
result_numpy = numpy_data + 5
end = time.time()
print("NumPy Array Time:", end - start)
```
